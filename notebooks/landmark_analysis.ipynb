# Imports

```python
import sys
import os
import yaml
import numpy as np

sys.path.append("../src")

from mediapipe_extraction.extract_landmarks import (
    process_source_a_sample,
    FEATURE_DIM,
)
```

# Load configuration

```python
with open("..\configs\dataset_paths.yaml", "r", encoding="utf-8") as f:
    config = yaml.safe_load(f)

profile = config["dataset"]["active_profile"]

main_root = config["dataset"]["source_a"][profile]["root"]
aux_root = config["dataset"]["source_b"][profile]["root"]

print("Main dataset:")
print(main_root)

print("\nAuxiliary dataset:")
print(aux_root)

print("\nExpected feature dimension:")
print(FEATURE_DIM)
```

# choose one real sample

```python
sample_path = os.path.join(
    main_root,
    "train",
    "Ambulance",
    "sample_001"
)

print(sample_path)
print(os.listdir(sample_path)[:10])
```

# extract landmarks

```python
landmarks, metadata = process_source_a_sample(
    sample_path
)

print("Metadata:")
print(metadata)

print("\nLandmark shape:")
print(landmarks.shape)

print("\nExpected feature dimension:")
print(FEATURE_DIM)
```

# check whether landmarks were actually detected

```python
print("Total values:", landmarks.size)

print("Non-zero values:", np.count_nonzero(landmarks))

print(
    "Non-zero percentage:",
    100 * np.count_nonzero(landmarks) / landmarks.size,
    "%"
)
```

```python
print("Min:", landmarks.min())
print("Max:", landmarks.max())
print("Mean:", landmarks.mean())
```
